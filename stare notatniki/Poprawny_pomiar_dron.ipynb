```python
# Importowanie najwazniejszych bibliotek
%matplotlib inline
import importlib

import matplotlib.pyplot as plt
import numpy as np

# Importowanie najwazniejszych komponentow Sionna RT
from sionna.rt import Camera, PathSolver
from sionna_env import setup_scene
import ofdm_config as cfg
importlib.reload(cfg)

my_cam = Camera(position=[-100, -140, 150.0],
                look_at=[-40, -20, 30])

scene_file = "./OpenStreetMap/kampus_AGH_doppler_drone/Kampus_AGH_B9_doppler_drone.xml"

scene, tx, rx = setup_scene(scene_file, cfg.fc, cfg.rx_rows, cfg.rx_cols)

dron = scene.get("dron_001")

p_solver = PathSolver()

# Konfiguracja ruchu
dron_velocity = np.array([10, -15, -2.0])  # prędkość drona w m/s (w dół)
dron.velocity = dron_velocity

displacement_vec = dron_velocity * 0.1
num_steps = 5

historia_taps = []

# Symulacja ruchu drona
for i in range(num_steps):
    
    paths = p_solver(scene=scene,
                 max_depth=3,
                 samples_per_src=500_000,
                 los=True,
                 specular_reflection=True,
                 diffuse_reflection=True,
                 refraction=True,
                 synthetic_array=False,
                 seed=67)

    a, tau = paths.taps(bandwidth=cfg.bandwidth,
                 num_time_steps=cfg.num_symbols,
                 sampling_frequency=cfg.ofdm_symbol_rate,
                 l_min = -6,
                 l_max = 100)

    historia_taps.append((a.numpy(), tau.numpy()))
    
    # Renderowanie Mapy - opcjonalnie można odkomentować poniższe linie, aby zobaczyć wizualizację sceny
    scene.render(camera=my_cam, 
                 paths=paths, 
                 rm_show_color_bar=True,
                 rm_vmax=-40, 
                 rm_vmin=-120,
                 resolution=(800, 600)) 


    # ruch drona
    dron.position = dron.position + displacement_vec

np.save("historia_taps_dron.npy", historia_taps, allow_pickle=True)
```
